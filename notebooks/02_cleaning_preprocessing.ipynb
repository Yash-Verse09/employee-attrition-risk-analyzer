# Cleaning and Preprocessing
Employee Attrition Risk Analyzer

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv")
print("Before:", df.shape)
```

```python

df = df.drop(columns=["EmployeeCount", "Over18", "StandardHours"])

# duplicates hatao aur index reset karo
df = df.drop_duplicates().reset_index(drop=True)

print("After:", df.shape)
print("Missing values:", df.isnull().sum().sum())
```

```python
df.to_csv("../data/processed/cleaned_data.csv", index=False)
```

```python
df_model = df.drop(columns=["EmployeeNumber"]).copy()

df_model["Attrition"] = df_model["Attrition"].map({"Yes": 1, "No": 0})

df_model["OverTime"] = df_model["OverTime"].map({"Yes": 1, "No": 0})
df_model["Gender"] = df_model["Gender"].map({"Male": 1, "Female": 0})

cat_cols = df_model.select_dtypes(include="object").columns.tolist()
print("One-hot columns:", cat_cols)

df_model = pd.get_dummies(df_model, columns=cat_cols, drop_first=True, dtype=int)
print(df_model.shape)
df_model.head()
```

```python
print("Text columns left:", df_model.select_dtypes(include="object").shape[1])
print("Missing:", df_model.isnull().sum().sum())
print(df_model["Attrition"].value_counts())

df_model.to_csv("../data/processed/model_ready_data.csv", index=False)
```
